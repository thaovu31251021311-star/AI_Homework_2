```python
#===== CÂU 1 ==================
class Investment:
    def __init__(self, principal, interest):
        self.principal = principal
        self.interest = interest

    def value_after(self, n):
        return self.principal * (1 + self.interest)**n

    def __str__(self):
        return f"Principal - ${self.principal:.2f}, Interest rate - {self.interest * 100:.2f}%"

investment = Investment(1000, 0.0512)

print(investment)
print(investment.value_after(5))
```

```python
#===== CÂU 2 ==================
class Product:
    def __init__(self, name, amount, price):
        self.name = name
        self.amount = amount
        self.price = price

    def get_price(self, quantity):
        if quantity < 10:
            discount = 0
        elif quantity > 10 and quantity < 100:
            discount = 0.10
        else:
            discount = 0.20
        return quantity * self.price * (1 - discount)

    def make_purchase(self, quantity):
        self.amount -= quantity

p = Product("Laptop", 200, 1000000)

print(p.get_price(5))
print(p.get_price(20))
print(p.get_price(100))

p.make_purchase(20)

print(p.amount)
```

```python
#===== CÂU 3 ==================
class Password_manager:
    def __init__(self, old_passwords):
        self.old_passwords = old_passwords

    def get_password(self):
        return self.old_passwords[-1]

    def set_password(self, new_password):
        if new_password not in self.old_passwords:
            self.old_passwords.append(new_password)

    def is_correct(self, password):
        return password == self.get_password()

p = Password_manager(["123vuthao", "1311.suize"])

print(p.get_password())
print(p.is_correct("1311.suize"))

p.set_password("vuthao_1311")

print(p.get_password())
print(p.old_passwords)
```

```python
#===== CÂU 4 ==================
class Time:
    def __init__(self, seconds):
        self.seconds = seconds

    def convert_to_minutes(self):
        minutes = self.seconds // 60
        seconds = self.seconds % 60
        return f"{minutes}:{seconds:02d}"

    def convert_to_hours(self):
        hours = self.seconds // 3600
        minutes = (self.seconds % 3600) // 60
        seconds = self.seconds % 60

        return f"{hours}:{minutes:02d}:{seconds:02d}"

t = Time(230)

print(t.convert_to_minutes())
print(t.convert_to_hours())
```

```python
#===== CÂU 5 ==================
class Wordplay:
    def __init__(self, words):
        self.words = words

    def words_with_length(self, length):
        return [word for word in self.words if len(word) == length]

    def starts_with(self, s):
        return [word for word in self.words if word.startswith(s)]

    def ends_with(self, s):
        return [word for word in self.words if word.endswith(s)]

    def palindromes(self):
        return [word for word in self.words if word == word[::-1]]

    def only(self, L):
        return [word for word in self.words
                if all(letter in L for letter in word)]

    def avoids(self, L):
        return [word for word in self.words
                if all(letter not in L for letter in word)]
words = ["logistics", "shipper", "capacity", "monopoly", "inbound", "outbound", "truckload"]

w = Wordplay(words)

print(w.words_with_length(7))
print(w.starts_with("lo"))
print(w.ends_with("ty"))
print(w.palindromes())
print(w.only(["m", "p", "l"]))
print(w.avoids(["a", "e"]))
```

```python
#===== CÂU 6 ==================
class Converter:
    def __init__(self, length, unit):
        self.length = length
        self.unit = unit

    def inches(self):
        return self.length * self.get_conversion_factor("inches")

    def feet(self):
        return self.length * self.get_conversion_factor("feet")

    def yards(self):
        return self.length * self.get_conversion_factor("yards")

    def miles(self):
        return self.length * self.get_conversion_factor("miles")

    def kilometers(self):
        return self.length * self.get_conversion_factor("kilometers")

    def meters(self):
        return self.length * self.get_conversion_factor("meters")

    def centimeters(self):
        return self.length * self.get_conversion_factor("centimeters")

    def millimeters(self):
        return self.length * self.get_conversion_factor("millimeters")

    def get_conversion_factor(self, target_unit):
        factors = {
            "inches": 1,
            "feet": 1 / 12,
            "yards": 1 / 36,
            "miles": 1 / 63360,
            "kilometers": 0.0000254,
            "meters": 0.0254,
            "centimeters": 2.54,
            "millimeters": 25.4
        }

        return factors[self.unit] / factors[target_unit]

c = Converter(9, "inches")

print("Inches:", c.inches())
print("Feet:", c.feet())
print("Yards:", c.yards())
print("Miles:", c.miles())
print("Kilometers:", c.kilometers())
print("Meters:", c.meters())
print("Centimeters:", c.centimeters())
print("Millimeters:", c.millimeters())
```

```python
#===== CÂU 9 ==================
import random

class Rock_paper_scissors:
    def __init__(self, rounds):
        self.rounds = rounds
        self.current_round = 0
        self.user_wins = 0
        self.computer_wins = 0

    def computer_choice(self):
        choices = ["rock", "paper", "scissors"]
        return random.choice(choices)

    def round_winner(self, user_choice, computer_choice):
        if user_choice == computer_choice:
            return "tie"

        if (
            (user_choice == "rock" and computer_choice == "scissors")
            or (user_choice == "paper" and computer_choice == "rock")
            or (user_choice == "scissors" and computer_choice == "paper")
        ):
            self.user_wins += 1
            return "user"

        self.computer_wins += 1
        return "computer"

    def game_winner(self):
        if self.user_wins > self.rounds // 2:
            return "user"

        if self.computer_wins > self.rounds // 2:
            return "computer"

        if self.current_round >= self.rounds:
            if self.user_wins > self.computer_wins:
                return "user"
            elif self.computer_wins > self.user_wins:
                return "computer"
            else:
                return "tie"

        return None

game = Rock_paper_scissors(5)

print(game.computer_choice())

print(game.round_winner("rock", "scissors"))
print(game.user_wins)

print(game.round_winner("paper", "scissors"))
print(game.computer_wins)
```
